# Tutorial 3: Reproducing All Manuscript Figures and Tables on Pure CPU

Every numerical table (Tables 1–4, Tables S1–S3) and every publication figure (Graphical Abstract, Figs. 1–9, Figs. S1–S5) in the manuscript is reproduced deterministically on CPU directly from the SHA-256-verified archive in `results_summary/`.

```python
import json
from pathlib import Path
from swi_tta.visualization.tables import generate_all_tables
from swi_tta.visualization.main_figures import make_graphical_abstract

REPO_ROOT = Path.cwd().parent if Path.cwd().name == 'tutorials' else Path.cwd()
SUMMARY_DIR = REPO_ROOT / 'results_summary'

cal = json.loads((SUMMARY_DIR / 'calibration_and_domains.json').read_text(encoding='utf-8'))
print('Calibrated domain noise floors (sigma_hat_D):')
for k, v in cal['calibrated_noise_floors'].items():
    ratio = f"{v['ratio_sigma_over_gt_eval']:.2f}x" if 'ratio_sigma_over_gt_eval' in v else 'field'
    print(f"  {k:<25s} ({v['regime']:<7s}): sigma_hat_D = {v['sigma_hat_d']:.5f} ({ratio})")
```

```python
# Regenerate all 7 Markdown + LaTeX tables and the Graphical Abstract into a temporary folder
out_root = Path('/tmp/swi_tta_tutorial_repro')
tbl_rec = generate_all_tables(str(SUMMARY_DIR), str(out_root / 'tables'))
ga_rec = make_graphical_abstract(str(SUMMARY_DIR), str(out_root / 'figures'), formats=('png',), dpi=150)
print(f'Generated {len(tbl_rec)} tables and Graphical Abstract at {out_root}')
print('To reproduce all 15 figures at 600 DPI (PDF/PNG/SVG), run: python scripts/reproduce_all_figures.py')
```
